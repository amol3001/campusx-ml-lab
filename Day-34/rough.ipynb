```python
import numpy as np
import pandas as pd
```

```python
date = pd.read_csv('date.csv')
```

```python
date.head()
```

```python
date.info()
```

# Converting data type of data column

```python
date['Date'] = pd.to_datetime(date['Date'])
```

```python
date.info()
```

# Creating seprate column for Year Month And Date

```python
date['Year']=date['Date'].dt.year
date['Month']=date['Date'].dt.month
date['Day']=date['Date'].dt.day
```

```python
date
```

# Cretaing seprate column for month and day name

```python
date['Month_Name'] = date['Date'].dt.month_name()
date['Day_Name'] = date['Date'].dt.day_name()
```

```python
date
```

# Day of week and is weekend columns

```python
date['DOW'] = date['Date'].dt.dayofweek
```

```python
date['Is_Weekend'] = np.where(date['Day_Name'].isin(['Saturday','Sunday']),1,9)
```

```python
date
```

# week column in context of year and qurater of year

```python
date['Week_of_year'] = date['Date'].dt.isocalendar().week
```

```python
date['Quarter_OF_Day'] = date['Date'].dt.quarter
```

```python
date
```

# datetime module for calculations

```python
import datetime
today = datetime.datetime.today()
print(today)
```

```python
date['Time_Since_Day']=today - date['Date']
```

```python
date
```

```python
(today - date['Date']).dt.days
```

```python
# Months passed

np.round((today - date['Date']) / np.timedelta64(1, 'D') / 30.44, 0)
```
