```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('encoding.csv')
```

```python
train = df[['Review','Education']]
test = df['Purchased']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(train,test,test_size=0.2,random_state=0)
```

```python
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import LabelEncoder
```

```python
oe = OrdinalEncoder()
le = LabelEncoder()
```

```python
X_train_scaled = oe.fit_transform(X_train)
X_test_scaled = oe.transform(X_test)
y_train_scaled = le.fit_transform(y_train)
y_test_scaled=le.transform(y_test)
```

```python
X_train_scaled = pd.DataFrame(X_train_scaled,columns=X_train.columns,dtype='int')
```

```python
X_test_scaled = pd.DataFrame(X_test_scaled,columns=X_test.columns,dtype='int')
```
