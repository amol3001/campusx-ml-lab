## Summary — Concepts Covered

| Concept | Formula / Key Point |
|---|---|
| Nominal data | No natural order — avoid Ordinal/Label Encoding, prefer One-Hot |
| Ordinal data | Has natural order — Ordinal Encoding fits well |
| OrdinalEncoder | For INPUT features, supports multiple columns, custom order via `categories=` |
| LabelEncoder | For the TARGET (y), single column, alphabetical order |
| Fit rule | Always fit on training data only, transform on test data |
| Unseen categories | Use `handle_unknown='use_encoded_value'` to avoid errors |
