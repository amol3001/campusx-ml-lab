# Lab Workbook: EDA using Univariate Analysis
This workbook explores univariate analysis methods for both categorical and numerical features using the Titanic dataset.

## Task 1: Analyzing Categorical Features
Extract frequency values, generate a count plot, and visualize a percentage pie chart for categorical features.

## Task 2: Analyzing Numerical Features and Outliers
Build histograms to inspect feature distributions, check for outliers using box plots, and calculate asymmetry measurements.
