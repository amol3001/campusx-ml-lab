# Lab Workbook: EDA using Univariate Analysis
This workbook explores univariate analysis methods for both categorical and numerical features using the Titanic dataset.

## Task 1: Analyzing Categorical Features
Extract frequency values, generate a count plot, and visualize a percentage pie chart for categorical features.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df = pd.read_csv(url)

# Categorical Frequencies and Distributions
print(df['Survived'].value_counts())
sns.countplot(data=df, x='Survived')
plt.show()

df['Pclass'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.show()
```

## Task 2: Analyzing Numerical Features and Outliers
Build histograms to inspect feature distributions, check for outliers using box plots, and calculate asymmetry measurements.

```python
# Numeric Continuous Range Evaluation
plt.hist(df['Age'].dropna(), bins=20)
plt.show()

# Outlier Auditing via Box Plot
sns.boxplot(x=df['Fare'])
plt.show()

# Balance & Skewness Check
print('Age Distribution Skewness:', df['Age'].skew())
print('Fare Distribution Skewness:', df['Fare'].skew())
```
