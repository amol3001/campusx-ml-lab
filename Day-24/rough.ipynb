```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Social_Network_Ads.csv')
```

```python
df.sample()
```

```python
df = df[['Age','EstimatedSalary','Purchased']]
```

```python
train = df[['Age','EstimatedSalary']]
```

```python
test = df['Purchased']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(train,test,test_size=0.2,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
scaler.fit(X_train)
```

```python
scaler.mean_
```

```python
X_train_scaled = scaler.transform(X_train)
```

```python
X_test_scaled = scaler.transform(X_test)
```

```python
X_test_scaled.shape
```

```python
X_train.describe()
```

```python
X_train_scaled =pd.DataFrame(X_train_scaled,columns=X_train.columns)
```

```python
X_test_scaled = pd.DataFrame(X_test_scaled,columns=X_test.columns)
```

```python
np.round(X_train_scaled.describe(),1)
```

```python
np.round(X_test_scaled.describe(),1)
```

```python
sns.scatterplot(data=X_train,x='Age',y='EstimatedSalary')
```

```python
sns.scatterplot(data=X_train_scaled,x='Age',y='EstimatedSalary')
```

```python
sns.scatterplot(data = X_test, x= 'Age',y='EstimatedSalary')
```

```python
sns.scatterplot(data=X_test_scaled,x='Age',y='EstimatedSalary')
```

#Why scaling is important

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression()
lr_scaled = LogisticRegression()
```

```python
lr.fit(X_test,y_test)
lr_scaled.fit(X_test_scaled,y_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
lr_pred = lr.predict(X_test)
lr_pred_scaled = lr_scaled.predict(X_test)
```

```python
print('Noraml;',accuracy_score(y_test,lr_pred))
print('Scaled:',accuracy_score(y_test,lr_pred_scaled))
```
