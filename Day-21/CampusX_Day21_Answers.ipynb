## ⚙️ Setup

```python
# Amol Jagtap | amoljagtap3001@gmail.com
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

tips    = sns.load_dataset('tips')
iris    = sns.load_dataset('iris')
titanic = sns.load_dataset('titanic')

sns.set_theme(style='whitegrid')
print('Ready!')
```

---
## Section 1: Dataset Exploration — ANSWERS

### 1.1 — Data Overview

```python
print(tips.head())
print('Shape:', tips.shape)
print(tips.dtypes)

numerical   = tips.select_dtypes(include='number').columns.tolist()
categorical = tips.select_dtypes(include='object').columns.tolist()
print('Numerical  :', numerical)
print('Categorical:', categorical)
# Numerical  : ['total_bill', 'tip', 'size']
# Categorical: ['sex', 'smoker', 'day', 'time']
```

### 1.2 — Missing Values

```python
print('Tips missing values:')
print(tips.isnull().sum())

print('\nTitanic missing values:')
print(titanic.isnull().sum())

most_missing = titanic.isnull().sum().idxmax()
print(f'\nMost missing values: {most_missing}')
# Answer: 'deck' has the most missing values
```

### 1.3 — Basic Stats

```python
print(tips.describe())
print('Mean tip        :', tips['tip'].mean().round(2))
print('Max total_bill  :', tips['total_bill'].max())
```

---
## Section 2: Numerical vs Numerical — ANSWERS

### 2.1 — Basic Scatter

```python
plt.figure(figsize=(7, 5))
sns.scatterplot(data=tips, x='total_bill', y='tip')
plt.title('Total Bill vs Tip')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.show()
# Positive trend: higher bills tend to get higher tips
```

### 2.2 — Scatter with Hue, Style, Size

```python
plt.figure(figsize=(9, 6))
sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='sex',
    style='smoker',
    size='size',
    sizes=(30, 200),
    alpha=0.8
)
plt.title('total_bill vs tip — grouped by sex, smoker, party size')
plt.show()
# Observation: males tend to have slightly higher bills and tips;
# non-smokers and smokers have similar tipping patterns.
```

### 2.3 — Correlation

```python
r1 = tips['total_bill'].corr(tips['tip'])
r2 = tips['total_bill'].corr(tips['size'])
print(f'total_bill vs tip  : r = {r1:.3f}  (moderate positive)')
print(f'total_bill vs size : r = {r2:.3f}  (moderate positive)')

print('\nFull correlation matrix:')
print(tips.corr(numeric_only=True).round(3))

# Highest correlation pair: total_bill and tip (r ≈ 0.68)
```

### 2.4 — Heatmap

```python
corr = tips[['total_bill', 'tip', 'size']].corr()

plt.figure(figsize=(6, 4))
sns.heatmap(
    corr,
    annot=True,
    fmt='.2f',
    cmap='Blues',
    linewidths=0.5,
    vmin=-1, vmax=1
)
plt.title('Correlation Heatmap — Tips Numerical Columns')
plt.tight_layout()
plt.show()
```

### 2.5 — Pair Plot (Iris)

```python
sns.pairplot(
    iris,
    hue='species',
    diag_kind='kde'
)
plt.suptitle('Iris Pair Plot', y=1.02)
plt.show()

# Observation: petal_length and petal_width show a very strong
# positive linear relationship (r ≈ 0.96). Setosa is clearly
# separated from versicolor and virginica in petal dimensions.
```

---
## Section 3: Numerical vs Categorical — ANSWERS

### 3.1 — Bar Plot

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

sns.barplot(data=tips, x='day', y='total_bill',
            palette='Set2', ax=axes[0])
axes[0].set_title('Average Total Bill per Day')

sns.barplot(data=tips, x='day', y='total_bill',
            hue='sex', palette='Set1', ax=axes[1])
axes[1].set_title('Average Total Bill per Day (by Sex)')

plt.tight_layout()
plt.show()

# Highest bill day: Sunday (Sun)
print('Average bill per day:')
print(tips.groupby('day')['total_bill'].mean().round(2))
```

### 3.2 — Box Plot

```python
plt.figure(figsize=(9, 5))
sns.boxplot(data=tips, x='day', y='total_bill',
            hue='sex', palette='pastel')
plt.title('Total Bill Distribution by Day and Sex')
plt.show()

# 3. Males have higher median bills than females on all days.
# 4. Males on Saturday have the most outliers.
```

### 3.3 — Violin Plot

```python
plt.figure(figsize=(10, 5))
sns.violinplot(
    data=tips,
    x='day',
    y='tip',
    hue='sex',
    split=True,
    palette='muted'
)
plt.title('Tip Distribution by Day and Sex (Split Violin)')
plt.show()

# Observation: Violin shows the full shape of the distribution —
# you can see bimodal shapes (two humps) that a box plot hides.
```

### 3.4 — Violin + Strip Overlay

```python
fig, ax = plt.subplots(figsize=(10, 5))

sns.violinplot(
    data=tips, x='day', y='total_bill',
    inner=None,          # remove inner box/quartile marks
    palette='muted',
    alpha=0.7,
    ax=ax
)

sns.stripplot(
    data=tips, x='day', y='total_bill',
    color='black',
    size=3,
    jitter=True,
    alpha=0.5,
    ax=ax
)

ax.set_title('Violin + Strip: total_bill by day')
plt.show()
```

## Section 4: Categorical vs Categorical — ANSWERS

### 4.1 — Cross-Tabulation

```python
# 1. Raw counts
ct_raw = pd.crosstab(tips['sex'], tips['day'])
print('Raw counts:')
print(ct_raw)

# 2. Normalised
ct_norm = pd.crosstab(tips['sex'], tips['day'],
                       normalize='index').round(3)
print('\nNormalised (row proportions):')
print(ct_norm)

# 3. Day with highest female proportion
highest_day = ct_norm.loc['Female'].idxmax()
print(f'\nHighest female proportion day: {highest_day}')
```

### 4.2 — Cross-Tab with Values

```python
mean_tip = pd.crosstab(
    tips['sex'],
    tips['day'],
    values=tips['tip'],
    aggfunc='mean'
).round(2)
print('Mean tip per sex × day:')
print(mean_tip)
```

### 4.3 — Grouped and Stacked Bar

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Grouped bar
sns.countplot(data=tips, x='day', hue='sex',
              palette='Set1', ax=axes[0])
axes[0].set_title('Count of Visits per Day by Sex')

# Stacked bar
ct = pd.crosstab(tips['day'], tips['sex'])
ct.plot(kind='bar', stacked=True,
        color=['steelblue', 'salmon'],
        ax=axes[1], rot=0)
axes[1].set_title('Stacked: Visits per Day by Sex')
axes[1].set_xlabel('Day')

plt.tight_layout()
plt.show()
```

### 4.4 — Categorical Heatmap

```python
ct = pd.crosstab(tips['sex'], tips['smoker'])

plt.figure(figsize=(5, 3))
sns.heatmap(ct,
            annot=True,
            fmt='d',
            cmap='YlOrRd',
            linewidths=0.5)
plt.title('Count: sex vs smoker')
plt.tight_layout()
plt.show()
```

## Section 5: Multivariate Analysis — ANSWERS

### 5.1 — 4-Variable Scatter

```python
plt.figure(figsize=(9, 6))
sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='sex',
    style='smoker',
    size='size',
    sizes=(30, 220),
    alpha=0.8
)
plt.title('4-Variable Scatter: total_bill vs tip')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

### 5.2 — FacetGrid

```python
g = sns.FacetGrid(
    tips,
    col='time',
    row='sex',
    height=3,
    aspect=1.2
)
g.map(sns.histplot, 'total_bill', bins=15, color='steelblue')
g.set_axis_labels('Total Bill ($)', 'Count')
g.set_titles(col_template='{col_name}', row_template='{row_name}')
plt.show()
```

### 5.3 — Pair Plot (Iris, specific columns)

```python
sns.pairplot(
    iris[['petal_length', 'petal_width', 'species']],
    hue='species',
    diag_kind='hist'
)
plt.show()

# Most separable species: Iris-setosa — it clusters far from
# versicolor and virginica in both petal dimensions.
```

### 5.4 — Titanic Lower-Triangle Heatmap

```python
t_num = titanic.select_dtypes(include='number').dropna()
corr  = t_num.corr()

mask = np.triu(np.ones_like(corr, dtype=bool))

plt.figure(figsize=(8, 6))
sns.heatmap(
    corr,
    mask=mask,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    linewidths=0.5,
    center=0
)
plt.title('Titanic Correlation Heatmap (Lower Triangle)')
plt.tight_layout()
plt.show()
```

## Section 6: Full EDA Challenge — ANSWERS

### Q1 — Fare vs Survived

```python
plt.figure(figsize=(7, 5))
sns.boxplot(
    data=titanic.dropna(subset=['fare']),
    x='survived', y='fare',
    palette=['salmon', 'steelblue']
)
plt.xticks([0, 1], ['Did Not Survive', 'Survived'])
plt.title('Fare Distribution by Survival')
plt.show()
# Observation: survivors paid significantly higher fares on median,
# suggesting wealthier passengers had better survival odds.
```

### Q2 — Class vs Survival

```python
plt.figure(figsize=(6, 4))
sns.barplot(
    data=titanic,
    x='pclass', y='survived',
    palette='Set2',
    errorbar=None
)
plt.title('Survival Rate by Passenger Class')
plt.ylabel('Survival Rate')
plt.xlabel('Passenger Class')
plt.show()
# Observation: 1st class passengers had ~63% survival rate vs ~24% for 3rd class.
```

### Q3 — Sex vs Survived

```python
ct = pd.crosstab(titanic['sex'], titanic['survived'],
                 colnames=['survived'])
print(ct)
print()
print(pd.crosstab(titanic['sex'], titanic['survived'],
                  normalize='index').round(3))

plt.figure(figsize=(5, 3))
sns.heatmap(ct, annot=True, fmt='d', cmap='Blues',
            linewidths=0.5)
plt.title('Count: sex vs survived')
plt.tight_layout()
plt.show()
# Observation: ~74% of females survived vs only ~19% of males.
```

### Q4 — Age vs Fare Scatter

```python
t_clean = titanic.dropna(subset=['age', 'fare'])

plt.figure(figsize=(9, 6))
sns.scatterplot(
    data=t_clean,
    x='age', y='fare',
    hue='survived',
    style='sex',
    alpha=0.6,
    palette=['salmon', 'steelblue']
)
plt.title('Age vs Fare (coloured by survival, shape by sex)')
plt.show()
# Observation: high-fare passengers (any age) tended to survive.
# No strong age-fare correlation visible.
```

### Q5 — Titanic EDA Dashboard

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Top-left: Age by survived
sns.boxplot(data=titanic.dropna(subset=['age']),
            x='survived', y='age',
            palette=['salmon', 'steelblue'], ax=axes[0, 0])
axes[0, 0].set_xticks([0, 1])
axes[0, 0].set_xticklabels(['Did Not Survive', 'Survived'])
axes[0, 0].set_title('Age by Survival')

# Top-right: Fare by pclass
sns.violinplot(data=titanic.dropna(subset=['fare']),
               x='pclass', y='fare',
               palette='Set2', ax=axes[0, 1])
axes[0, 1].set_title('Fare Distribution by Class')

# Bottom-left: Survival count by sex
sns.countplot(data=titanic, x='sex',
              hue='survived',
              palette=['salmon', 'steelblue'],
              ax=axes[1, 0])
axes[1, 0].set_title('Survival Count by Sex')
axes[1, 0].legend(title='Survived', labels=['No', 'Yes'])

# Bottom-right: Correlation heatmap
t_n = titanic.select_dtypes(include='number').dropna()
sns.heatmap(t_n.corr(),
            annot=True, fmt='.2f',
            cmap='coolwarm', center=0,
            linewidths=0.5,
            ax=axes[1, 1])
axes[1, 1].set_title('Correlation Matrix')

plt.suptitle('Titanic EDA Dashboard — Day 21',
             fontsize=15, fontweight='bold')
plt.tight_layout()
plt.savefig(
    '/mnt/user-data/outputs/campusx_day21/titanic_dashboard.png',
    dpi=150, bbox_inches='tight'
)
plt.show()
print('Dashboard saved!')
```
