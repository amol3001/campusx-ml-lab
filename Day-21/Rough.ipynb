```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset('tips')
```

```python
flights = sns.load_dataset('flights')
```

```python
iris = sns.load_dataset('iris')
```

```python
titanic = pd.read_csv('titanic.csv')
```

```python
tips
```

# 1) Scatter plot

```python
sns.set_style('dark')
#sns.set_palette('Set1')
color = ['black','orange']
sns.scatterplot(data = tips,x = 'total_bill',y = 'tip',hue='sex',style='smoker',palette=color,size='size')
```

# 2)Barplot

```python
titanic
```

```python
sns.barplot(data = titanic,x = 'Survived',y = 'Fare')
```

```python
sns.barplot(data = titanic, x = 'Pclass', y = 'Fare',palette=color,hue='Sex')
```

# 3) Boxplot

```python
sns.boxplot(data = titanic,x = 'Sex',y='Age',palette=color,hue = 'Survived')
```

```python
sns.distplot(titanic['Age'],kde=True,hist=False)
```

```python
sns.heatmap(pd.crosstab(titanic['Pclass'],titanic['Survived']))
```

```python
sns.heatmap(data = titanic,x = 'Embarked',y = 'Embarked')
```

# 6) Cluster map

```python
sns.clustermap(pd.crosstab(titanic['Parch'],titanic['Survived']))
```

# 7) Pairplot

```python
sns.pairplot(iris,hue='species')
```
